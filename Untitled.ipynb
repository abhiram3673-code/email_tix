```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
model_cat=LogisticRegression(max_iter=1000)
model_cat.fit(x_train,y_cat_train)
cat_pred=model_cat.predict(x_test)
acc=accuracy_score(y_cat_test,cat_pred)
acc
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_cat_test, model_cat.predict(x_test)))
```

```python

model_pri=LogisticRegression(max_iter=1000)
model_pri.fit(x_train,y_prio_train)
prio_pred=model_pri.predict(x_test)
acc=accuracy_score(y_prio_test,prio_pred)
acc
print(classification_report(y_prio_test, model_pri.predict(x_test)))
```
